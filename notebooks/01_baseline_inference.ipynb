```python
# Baseline inference: pretrained COCO model, no training
from ultralytics import YOLO
import glob
from IPython.display import Image, display
model = YOLO("yolov8n.pt")
imgs = sorted(glob.glob("/content/repo/new_images/*"))[:5]
for p in imgs:
    r = model.predict(p, conf=0.25, save=True, project="/content/baseline", name="pred", exist_ok=True, verbose=False)[0]
    print(os.path.basename(p), "->", len(r.boxes), "detections:", [model.names[int(c)] for c in r.boxes.cls])
for p in sorted(glob.glob("/content/baseline/pred/*"))[:5]:
    display(Image(p, width=600))
```
