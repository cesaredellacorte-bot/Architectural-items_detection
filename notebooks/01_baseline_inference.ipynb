```
# Baseline inference: pretrained COCO model, no training -- shows what a generic model does on AECO images
!pip install -q ultralytics
from ultralytics import YOLO
import glob
from IPython.display import Image, display
model = YOLO("yolov8n.pt")
imgs = sorted(glob.glob("/content/repo/new_images/*"))[:5]   # upload some images to /new_images in the repo
for p in imgs:
    r = model.predict(p, conf=0.25, save=True, project="/content/baseline", name="pred", exist_ok=True, verbose=False)[0]
    print(p, "->", len(r.boxes), "detections")
for p in sorted(glob.glob("/content/baseline/pred/*"))[:5]:
    display(Image(p, width=600))
```
